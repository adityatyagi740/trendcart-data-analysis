🎯 Your Mini-Challenge for Task 2:
Write Python code using Pandas to perform these cleaning steps:

Remove Duplicates:

Drop the exact duplicate rows from df and reset the DataFrame index (drop_duplicates()).

Handle Missing Values:

Price: Fill missing values with the median price of the dataset (or category-wise median).

Payment_Method: Fill missing values with the mode (most common payment method) or the string 'Unknown'.

Rating: Fill missing values with the mean rating (rounded to 1 decimal place).

Fix Data Types:

Convert the Date column into a Pandas datetime format using pd.to_datetime().

Verify Cleanliness:

Run df.isnull().sum() and df.duplicated().sum() to confirm all zeros.

```python
df.drop_duplicates(inplace=True)
```

```python
df.columns
```

```python
df["Price"]=df["Price"].fillna(df["Price"].median())
```

```python
df["Payment_Method"]=df["Payment_Method"].fillna(df["Payment_Method"].mode())
```

```python
df["Rating"]=df["Rating"].fillna(round(df["Rating"].mean(),1))
```
